```python
spark.sql("CREATE SCHEMA IF NOT EXISTS bronze")
```

```python
df = spark.read \
    .option("header", True) \
    .csv(raw_path)

df = df.withColumn(
        "ingestion_timestamp",
        current_timestamp()
    ) \
    .withColumn(
        "source_file",
        col("_metadata.file_path")
    ) \
    .withColumn(
        "load_date",
        current_date()
    )


```

```python
bronze_path = "abfss://bronze@stautofulfildata.dfs.core.windows.net/orders/"

(df.write
  .format("delta")
  .mode("append")
  .partitionBy("load_date")
  .save(bronze_path))
```
