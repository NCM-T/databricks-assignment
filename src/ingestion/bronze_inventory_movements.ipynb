```python
spark.sql("CREATE SCHEMA IF NOT EXISTS bronze")
```

```python
df = spark.read \
    .option("header", True) \
    .csv(raw_path)

df = df.withColumn(
        "ingestion_timestamp",
        current_timestamp()
    ) \
    .withColumn(
        "source_file",
        col("_metadata.file_path")
    ) \
    .withColumn(
        "load_date",
        current_date()
    )

(df.write
 .format("delta")
 .mode("append")
 .partitionBy("load_date")
 .saveAsTable("bronze.inventory_movements"))
```
